This notebook analyses two weeks of sales and the current stock records for Wanjiku's fresh-produce stall at Gikomba Market. The aim is to understand sales performance, compare the two weeks, classify daily performance and identify products that need attention.

## Task 1: Daily Sales Summary and Performance Analysis

I will calculate the total and average sales for the 14 days, then identify the highest and lowest sales amounts. I will use the positions of those amounts in the sales list to find the corresponding day names and determine whether each occurred in Week 1 or Week 2.

```python
# Sales and stock records provided in the assignment

daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of the highest and lowest sales in the list.
best_day_position = daily_sales.index(highest_sales)
worst_day_position = daily_sales.index(lowest_sales)

best_day = days[best_day_position]
worst_day = days[worst_day_position]
best_week = best_day_position // 7 + 1
worst_week = worst_day_position // 7 + 1

print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest daily sales: KES {highest_sales:,.2f} ({best_day}, Week {best_week})")
print(f"Lowest daily sales: KES {lowest_sales:,.2f} ({worst_day}, Week {worst_week})")
```

## Task 2: Weekly Sales Totals and Growth

I will define a function that uses list slicing to separate the first seven days from the last seven days and return their totals as a tuple. I will then unpack the two totals and calculate the percentage change from Week 1 to Week 2. A positive result means sales increased, while a negative result means they decreased.

```python
def week_totals(sales):
    """Return the total sales for the first and second seven-day periods."""
    first_week_total = sum(sales[:7])
    second_week_total = sum(sales[7:14])
    return first_week_total, second_week_total


week_1_total, week_2_total = week_totals(daily_sales)
growth_percentage = ((week_2_total - week_1_total) / week_1_total) * 100

print(f"Week 1 sales: KES {week_1_total:,.2f}")
print(f"Week 2 sales: KES {week_2_total:,.2f}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

## Task 3: Daily Sales Classification

I will classify each day's sales by comparing it with the overall daily average. Sales more than 20% above the average will be marked as **Strong**, while sales more than 20% below the average will be marked as **Weak**. Any sales between those thresholds will be marked as **Normal**. I will then count how many days fall into each category.

```python
def classify_day(amount, average):
    """Classify a day's sales as Strong, Normal or Weak against the average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Keep the classification for each day, then count each category.
daily_categories = []
category_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_categories.append(category)
    category_counts[category] += 1

for i in range(len(daily_sales)):
    print(f"{days[i]}: KES {daily_sales[i]:,.2f} - {daily_categories[i]}")

print("\nNumber of days in each category:")
for category, count in category_counts.items():
    print(f"{category}: {count}")
```

## Task 4: Weekday and Weekend Sales Comparison

I will separate the sales into weekdays (Monday to Friday) and weekends (Saturday and Sunday) using the day labels. I will calculate the average sales for each group and compare them using the weekday average as the baseline. This will show whether the stall typically earns more per day during the week or over the weekend.

```python
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)
weekend_difference = ((weekend_average - weekday_average) / weekday_average) * 100

print(f"Average weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")
print(f"Weekend average compared with weekday average: {weekend_difference:.2f}% higher")
```

## Task 5: Stock Reordering and Gross Margin

I will go through the stock dictionary and compare each product's current quantity with its reorder level. Products at or below that level will be added to a restock list. I will also calculate the gross margin percentage for every product using its selling price and cost price.

```python
def products_to_reorder(stock_records):
    """Return product names where current stock is at or below the reorder level."""
    reorder_items = []

    for product_name, details in stock_records.items():
        if details["qty"] <= details["reorder"]:
            reorder_items.append(product_name)

    return reorder_items


reorder_items = products_to_reorder(stock)

print("Products to reorder:")
for product_name in reorder_items:
    print(f"- {product_name}")

print("\nGross margin by product:")
for product_name, details in stock.items():
    margin_percentage = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product_name}: {margin_percentage:.2f}%")
```

## Task 6: Stall Performance Report

The report will show the overall sales figures, the weekly comparison, the weekday and weekend averages, the daily performance counts, the products that need restocking and the gross margins. ```python

```python
print("           STALL PERFORMANCE REPORT")

print("\nSALES SUMMARY")
print(f"Total sales (14 days):       KES {total_sales:>12,.2f}")
print(f"Average daily sales:         KES {average_sales:>12,.2f}")
print(f"Highest sales:               KES {highest_sales:>12,.2f} - {best_day}, Week {best_week}")
print(f"Lowest sales:                KES {lowest_sales:>12,.2f} - {worst_day}, Week {worst_week}")

print("\nWEEKLY PERFORMANCE")
print(f"Week 1 total:                KES {week_1_total:>12,.2f}")
print(f"Week 2 total:                KES {week_2_total:>12,.2f}")
print(f"Week-on-week growth:         {growth_percentage:>11.2f}%")

print("\nAVERAGE SALES BY DAY TYPE")
print(f"Weekday average:             KES {weekday_average:>12,.2f}")
print(f"Weekend average:             KES {weekend_average:>12,.2f}")
print(f"Weekend difference:          {weekend_difference:>11.2f}% higher")

print("\nDAILY PERFORMANCE")
for category, count in category_counts.items():
    print(f"{category:<12}: {count} day(s)")

print("\nSTOCK TO REORDER")
for product_name in reorder_items:
    print(f"- {product_name}")

print("\nGROSS MARGIN")
for product_name, details in stock.items():
    margin_percentage = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product_name:<12}: {margin_percentage:>6.2f}%")
```
